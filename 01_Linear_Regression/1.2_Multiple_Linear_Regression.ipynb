## Multiple Linear Regression

```python
# Import necessary libraries for data generation, manipulation, visualization, and evaluation
import numpy as np
import pandas as pd
from sklearn.datasets import make_regression


import plotly.express as px
import plotly.graph_objects as go

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Generate a synthetic 3D regression dataset with 100 samples and 2 features
X, y = make_regression(n_samples=100, n_features=2, n_informative=2, n_targets=1, noise=50)
```

```python
# Create a pandas DataFrame containing features and the target variable
df = pd.DataFrame({'feature1':X[:,0],'feature2':X[:,1],'target':y})
```

```python
### Check the dimensions (rows, columns) of the generated DataFrame
df.shape
```

```python
# Create and display a 3D scatter plot of our dataset using Plotly Express
fig = px.scatter_3d(df, x='feature1', y='feature2', z='target')
fig.show()
```

```python
# Split the generated data into training (80%) and testing (20%) sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=3)
```

```python
# Import the LinearRegression model class from scikit-learn
from sklearn.linear_model import LinearRegression
```

```python
# Instantiate the linear regression model object
lr = LinearRegression()
```

```python
# Fit (train) the Linear Regression model on the training data
lr.fit(X_train, y_train)
```

```python
# Use the trained model to predict the target values for the test set
y_pred = lr.predict(X_test)
```

```python
# Calculate and print the performance metrics (MAE, MSE, and R2 Score) on the test predictions
print('MAE ->', mean_absolute_error(y_test, y_pred))
print('MSE ->', mean_squared_error(y_test, y_pred))
print('R2 Score ->', r2_score(y_test, y_pred))
```

```python
# Set up coordinate grids and predict predictions across the feature space to draw the regression plane
x = np.linspace(-5, 5, 10)
y = np.linspace(-5, 5, 10)

xGrid, yGrid = np.meshgrid(y, x)

final = np.vstack((xGrid.ravel().reshape(1,100), yGrid.ravel().reshape(1,100))).T

z_final = lr.predict(final).reshape(10,10)

z = z_final
```

```python
# Plot the original data points and superimpose the 3D regression decision boundary plane
fig = px.scatter_3d(df, x='feature1', y='feature2', z='target')

fig.add_trace(go.Surface(x = x, y = y, z = z))

fig.show()
```

```python
# Retrieve the calculated slope coefficients for each feature
lr.coef_
```

```python
# Retrieve the calculated y-intercept value of the regression line
lr.intercept_
```

### Making our own Linear Regression Class

```python
# Define a custom Multiple Linear Regression class using the Closed-Form OLS Formulation
class MeraLR:

    def __init__(self):
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        # Insert a column of 1s at index 0 for the intercept/bias term
        X_train = np.insert(X_train, 0, 1, axis=1)

        # Solve for coefficients using normal equation: (X^T * X)^(-1) * X^T * y
        betas = np.linalg.inv(np.dot(X_train.T, X_train)).dot(X_train.T).dot(y_train)
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]

    def predict(self, X_test):
        # Compute the dot product of test data and coefficients, then add the intercept
        y_pred = np.dot(X_test, self.coef_) + self.intercept_
        return y_pred
```

```python
# Instantiate our custom MeraLR class object
lr = MeraLR()
```

```python
# Train our custom closed-form Multiple Linear Regression model on training data
lr.fit(X_train, y_train)
```

```python
# Preview the shape after prepend/insertion of bias column filled with ones
np.insert(X_train, 0, 1, axis=1).shape
```

```python
# Generate targets predictions using our custom model on test subset data
y_pred = lr.predict(X_test)
```

```python
# Evaluate performance by computing R2 score for custom implementation predictions
r2_score(y_test,y_pred)
```
